```python
### 1) Pearson and Spearman Correlation

# Libraries
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy as sp
from sklearn.metrics import r2_score

# Read the dataset csv file
df = pd.read_csv("/blue/env3040/jessicazhang1/Assignments/Li_et_al_2023_Gas_Exchange_Sheet.csv")   

#List of Relevant Variables
AOT40_ambient = list(df["AOT 40 - Ambient"]) 
AOT40_elevated = list(df["AOT 40 - Elevated"])
NPR_ambient = list(df["NPR - Ambient (umol m-2 s-1)"])
NPR_elevated = list(df["NPR - Elevated (umol m-2 s-1)"])
SC_ambient = list(df["SC - Ambient (mol m-2 s-1)"])
SC_elevated = list(df["SC - Elevated (mol m-2 s-1)"])
CO2_ambient = list(df["CO2 - Ambient (umol mol-1)"])
CO2_elevated = list(df["CO2 - Elevated (umol mol-1)"])


p_npr_a = sp.stats.pearsonr(AOT40_ambient, NPR_ambient)   # Calculate Pearson statistics for AOT40 and net photosynthesis rate - ambient
p_npr_e = sp.stats.pearsonr(AOT40_elevated, NPR_elevated)   # Calculate Pearson statistics for AOT40 and NPR - elevated

p_sc_a = sp.stats.pearsonr(AOT40_ambient, SC_ambient)   # Calculate Pearson statistics for AOT40 and stomatal conductance - ambient
p_sc_e = sp.stats.pearsonr(AOT40_elevated, SC_elevated)   # Calculate Pearson statistics for AOT40 and SC - elevated

p_co_a = sp.stats.pearsonr(AOT40_ambient, CO2_ambient)   # Calculate Pearson Statistics for AOT40 and Interal CO2 Concentrations - ambient
p_co_e = sp.stats.pearsonr(AOT40_elevated, CO2_elevated)   # Calculate Pearson Statistics for AOT40 and CO2 - elevated


s_npr_a = sp.stats.spearmanr(AOT40_ambient,  NPR_ambient)   # Calculate Spearman statistics for AOT40 and net photosynthesis rate - ambient
s_npr_e = sp.stats.spearmanr(AOT40_elevated, NPR_elevated)   # Calculate Spearman statistics for AOT40 and NPR - elevated

s_sc_a = sp.stats.spearmanr(AOT40_ambient, SC_ambient)   # Calculate Spearman statistics for AOT40 and stomatal conductance - ambient
s_sc_e = sp.stats.spearmanr(AOT40_elevated, SC_elevated)   # Calculate Spearman statistics for AOT40 and SC - elevated

s_co_a = sp.stats.spearmanr(AOT40_ambient, CO2_ambient)   # Calculate Spearman Statistics for AOT40 and Intercellular CO2 Concentrations - ambient
s_co_e = sp.stats.spearmanr(AOT40_elevated, CO2_elevated)   # Calculate Spearman Statistics for AOT40 and CO2 - elevated


# Return Pearson Correlation Coefficients for Relevant Varaiables
print("Pearson Correlation Coefficient: \n")
print("Ambient Ozone Exposure and Net Photosynthesis Rate:", p_npr_a, "\n")
print("Elevated Ozone Exposure and Net Photosynthesis Rate:", p_npr_e, "\n")
print("Ambient Ozone Exposure and Stomatal Conductance:", p_sc_a, "\n")
print("Elevated Ozone Exposure and Stomatal Conductance:", p_sc_e, "\n")
print("Ambient Ozone Exposure and Intercellular CO2:", p_co_a, "\n")
print("Elevated Ozone Exposure and Intercellular CO2:", p_co_e, "\n")

# Return Spearman Correlation Coefficients for Relevant Varaiables
print("\nSpearman Correlation Coefficients: \n")
print("Ambient Ozone Exposure and Net Photosynthesis Rate:", s_npr_a, "\n")
print("Elevated Ozone Exposure and Net Photosynthesis Rate:", s_npr_e, "\n")
print("Ambient Ozone Exposure and Stomatal Conductance:", s_sc_a, "\n")
print("Elevated Ozone Exposure and Stomatal Conductance:", s_sc_e, "\n")
print("Ambient Ozone Exposure and Intercellular CO2:", s_co_a, "\n")
print("Elevated Ozone Exposure and Intercellular CO2:", s_co_e)
```

```python
### 2) Linear Regression 

lr_npr_a = sp.stats.linregress(AOT40_ambient, NPR_ambient)   # Calculate Linear Regression for AOT40 and net photosynthesis rate - ambient
lr_npr_e = sp.stats.linregress(AOT40_elevated, NPR_elevated)   # Calculate Linear Regression for AOT40 and NPR - elevated

lr_sc_a = sp.stats.linregress(AOT40_ambient, SC_ambient)   # Calculate Linear Regression for AOT40 and stomatal conductance - ambient
lr_sc_e = sp.stats.linregress(AOT40_elevated, SC_elevated)   # Calculate Linear Regression for AOT40 and SC - elevated

lr_co_a = sp.stats.linregress(AOT40_ambient, CO2_ambient)   # Calculate Linear Regression for AOT40 and Intercellular CO2 Concentrations - ambient
lr_co_e = sp.stats.linregress(AOT40_elevated, CO2_elevated)   # Calculate Linear Regression for AOT40 and CO2 - elevated

# Return Linear Regression for Relevant Variables
print("Linear Regression: \n")
print("Ambient Ozone Exposure and Net Photosynthesis Rate:", lr_npr_a, "\n")
print("Elevated Ozone Exposure and Net Photosynthesis Rate:", lr_npr_e, "\n")
print("Ambient Ozone Exposure and Stomatal Conductance:", lr_sc_a, "\n")
print("Elevated Ozone Exposure and Stomatal Conductance:", lr_sc_e, "\n")
print("Ambient Ozone Exposure and Intercellular CO2:", lr_co_a, "\n")
print("Elevated Ozone Exposure and Intercellular CO2:", lr_co_e)
```

```python
### 3) Scatter Plot With Linear Regression Line

# Net Photosynthesis Rate vs Ozone Exposure - Ambient Scatter Plot 
x = (0, 4.5, 0.5)   # x-axis starts from 0 to 4.5 with interval of 0.5
y = []            

for i in x:   # Goes through x values
    j = lr_npr_a.slope * i + lr_npr_a.intercept  # Calculates the y values
    y.append(j)    # Adds the y values into the y list

    
# Create a scatter plot with linear regression line
fig, ax = plt.subplots()   # Creates an empty plot

ax.scatter(AOT40_ambient, NPR_ambient)   # Generates a scatter plot using the AOT40 and NPR ambient data values
ax.plot(x,y, color="orange")    # Generates a line graph using the x and y values
ax.minorticks_on()    # Add minor ticks on both x-axis and y-axis

ax.set_title("Net Photosynthesis Rate vs Ozone Exposure - Ambient")   # Names the graph
ax.set_xlabel("Ozone Exposure (ppb)")   # Names the x-axis
ax.set_ylabel("Net Photosynthesis Rate (umol m-2 s-1)")    # Names the y-axis
ax.legend(["Data", "Linear Regression"])   # Add legend that indicates which is the data or linear regression


# Net Photosynthesis Rate vs Ozone Exposure - Elevated Scatter Plot 
# Repeats the process

x2 = (0, 30, 5)
y2 = []

for k in x2:
    n = lr_npr_e.slope * k + lr_npr_e.intercept
    y2.append(n)

fig, ax2 = plt.subplots()

ax2.scatter(AOT40_elevated, NPR_elevated)
ax2.plot(x2,y2, color="orange")
ax2.minorticks_on()

# Create a variable to hold all of the Pearson and Spearman statistics
stats = """  
    Pearson r = -0.54,
    Pearson p = 5.26E-18,
    Spearman rho = -0.52,
    Spearman p = 4.09E-17,
"""

# Create a text block at the bottom left corner of the Pearson and Spearman statistics
ax2.text(
    0, 0.21,
    stats,
    verticalalignment="top",
    transform=ax.transAxes,
    fontsize=8,
    bbox=dict(
        boxstyle="round",
        facecolor="white",
        alpha=0.83)
)

ax2.set_title("Net Photosynthesis Rate vs Ozone Exposure - Elevated")
ax2.set_xlabel("Ozone Exposure (ppb)")
ax2.set_ylabel("Net Photosynthesis Rate (umol m-2 s-1)")
ax2.legend(["Data", "Linear Regression"])


# Stomatal Conductance vs Ozone Exposure - Ambient Scatter Plot 
x3 = (0, 4.5, 0.5)
y3 = []

for m in x3:
    p = lr_sc_a.slope * m + lr_sc_a.intercept
    y3.append(p)

fig, ax3 = plt.subplots()

ax3.scatter(AOT40_ambient, SC_ambient)
ax3.plot(x3,y3, color="orange")
ax3.minorticks_on()

ax3.set_title("Stomatal Conductance vs Ozone Exposure - Ambient")
ax3.set_xlabel("Ozone Exposure (ppb)")
ax3.set_ylabel("Stomata Conductance (mol m-2 s-1)")
ax3.legend(["Data", "Linear Regression"])


# Stomatal Conductance vs Ozone Exposure - Elevated Scatter Plot 
# Repeats the process

x4 = (0, 30, 5)
y4 = []

for q in x4:
    r = lr_sc_e.slope * q + lr_sc_e.intercept
    y4.append(r)

fig, ax4 = plt.subplots()

ax4.scatter(AOT40_elevated, SC_elevated)
ax4.plot(x4,y4, color="orange")
ax4.minorticks_on()

ax4.set_title("Stomatal Conductance vs Ozone Exposure - Elevated")
ax4.set_xlabel("Ozone Exposure (ppb)")
ax4.set_ylabel("Stomata Conductance (mol m-2 s-1)")
ax4.legend(["Data", "Linear Regression"])


# Intercellular CO2 Concentrations vs Ozone Exposure - Ambient Scatter Plot 
# Repeats the process

x5 = (0, 4.6, 0.5)
y5 = []

for s in x5:
    t = lr_co_a.slope * s + lr_co_a.intercept
    y5.append(t)

fig, ax5 = plt.subplots()

ax5.scatter(AOT40_ambient, CO2_ambient)
ax5.plot(x5,y5, color="orange")
ax5.minorticks_on()

ax5.set_title("Intercellular CO2 vs Ozone Exposure - Ambient")
ax5.set_xlabel("Ozone Exposure (ppb)")
ax5.set_ylabel("Intercellular CO2 Concentrations (umol mol-1)")
ax5.legend(["Data", "Linear Regression"])


# Intercellular CO2 Concentrations vs Ozone Exposure - Elevated Scatter Plot 
x6 = (0, 30, 5)
y6 = []

for u in x6:
    w = lr_co_e.slope * u + lr_co_e.intercept
    y6.append(w)

fig, ax6 = plt.subplots()

ax6.scatter(AOT40_elevated, CO2_elevated)
ax6.plot(x6,y6, color="orange")
ax6.minorticks_on()

ax6.set_title("Intercellular CO2 vs Ozone Exposure - Elevated")
ax6.set_xlabel("Ozone Exposure (ppb)")
ax6.set_ylabel("Intercellular CO2 Concentrations (umol mol-1)")
ax6.legend(["Data", "Linear Regression"])
```
