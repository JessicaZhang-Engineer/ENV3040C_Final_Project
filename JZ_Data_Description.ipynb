```python
### 1) Means, Medians, and Standard Deviations

# Libraries
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np

# Use pandas to read the dataset csv file
df = pd.read_csv("/blue/env3040/jessicazhang1/Assignments/Li_et_al_2023_Gas_Exchange_Sheet.csv") 

# Create a variable containing the specific column's values
AOT40_ambient = df["AOT 40 - Ambient"]
AOT40_elevated = df["AOT 40 - Elevated"]
NPR_ambient = df["NPR - Ambient (umol m-2 s-1)"]
NPR_elevated = df["NPR - Elevated (umol m-2 s-1)"]
SC_ambient = df["SC - Ambient (mol m-2 s-1)"]
SC_elevated = df["SC - Elevated (mol m-2 s-1)"]
CO2_ambient = df["CO2 - Ambient (umol mol-1)"]
CO2_elevated = df["CO2 - Elevated (umol mol-1)"]

# Mean, Median, and Standard Deviation of AOT40 - Ambient
AOT40_a_mean = AOT40_ambient.mean()
AOT40_a_median = AOT40_ambient.median()
AOT40_a_std = AOT40_ambient.std()

# Mean, Median, and Standard Deviation of AOT40 - Elevated
AOT40_e_mean = AOT40_elevated.mean()
AOT40_e_median = AOT40_elevated.median()
AOT40_e_std = AOT40_elevated.std()

# Mean, Median, and Standard Deviation of NPR - Ambient
NPR_a_mean = NPR_ambient.mean()
NPR_a_median = NPR_ambient.median()
NPR_a_std = NPR_ambient.std()

# Mean, Median, and Standard Deviation of NPR - Elevated
NPR_e_mean = NPR_elevated.mean()
NPR_e_median = NPR_elevated.median()
NPR_e_std = NPR_elevated.std()

# Mean, Median, and Standard Deviation of SC - Ambient
SC_a_mean = SC_ambient.mean()
SC_a_median = SC_ambient.median()
SC_a_std = SC_ambient.std()

# Mean, Median, and Standard Deviation of SC - Elevated
SC_e_mean = SC_elevated.mean()
SC_e_median = SC_elevated.median()
SC_e_std = SC_elevated.std()

# Mean, Median, and Standard Deviation of CO2 - Ambient
CO2_a_mean = CO2_ambient.mean()
CO2_a_median = CO2_ambient.median()
CO2_a_std = CO2_ambient.std()

# Mean, Median, and Standard Deviation of CO2 - Elevated
CO2_e_mean = CO2_elevated.mean()
CO2_e_median = CO2_elevated.median()
CO2_e_std = CO2_elevated.std()

# Return each Means, Medians, and Standard Deviations for Ambient and Elevated Variables
print("Ozone Exposure")
print("Ambient Ozone Exposure Mean:", AOT40_a_mean)
print("Ambient Ozone Exposure Median:", AOT40_a_median)
print("Ambient Ozone Exposure Standard Deviation:",AOT40_a_std, "\n")

print("Elevated Ozone Exposure Mean:", AOT40_e_mean)
print("Elevated Ozone Exposure Median:", AOT40_e_median)
print("Elevated Ozone Exposure Standard Deviation:", AOT40_e_std, "\n")


print("\nNet Photosynethesis Rate")
print("Ambient Net Photosynethesis Rate Mean:", NPR_a_mean)
print("Ambient Net Photosynethesis Rate Median:", NPR_a_median)
print("Ambient Net Photosynethesis Rate Standard Deviation:", NPR_a_std, "\n")

print("Elevated Net Photosynethesis Rate Mean:", NPR_e_mean)
print("Elevated Net Photosynethesis Rate Median:", NPR_e_median)
print("Elevated Net Photosynethesis Rate Standard Deviation::", NPR_e_std, "\n")


print("\nStomatal Conductance")
print("Ambient Stomatal Conductance Mean:", SC_a_mean)
print("Ambient Stomatal Conductance Median:", SC_a_median)
print("Ambient Stomatal Conductance Standard Deviation:", SC_a_std, "\n")

print("Elevated Stomatal Conductance Mean:", SC_e_mean)
print("Elevated Stomatal Conductance Median:", SC_e_median)
print("Elevated Stomatal Conductance Standard Deviation:", SC_e_std, "\n")


print("\nInternal CO2 Concentrations")
print("Ambient Intercellular CO2 Concentrations Mean:", CO2_a_mean)
print("Ambient Intercellular CO2 Concentrations Median:", CO2_a_median)
print("Ambient Intercellular CO2 Concentrations Standard Deviation:", CO2_a_std, "\n")

print("Elevated Intercellular CO2 Concentrations Mean:", CO2_e_mean)
print("Elevated Intercellular CO2 Concentrations Median:", CO2_e_median)
print("Elevated Intercellular CO2 Concentrations Standard Deviation:", CO2_e_std)
```

```python
### 2) Histograms of Relevant Variables

#List of Relevant Variables
AOT40_ambient = list(df["AOT 40 - Ambient"]) 
AOT40_elevated = list(df["AOT 40 - Elevated"])
NPR_ambient = list(df["NPR - Ambient (umol m-2 s-1)"])
NPR_elevated = list(df["NPR - Elevated (umol m-2 s-1)"])
SC_ambient = list(df["SC - Ambient (mol m-2 s-1)"])
SC_elevated = list(df["SC - Elevated (mol m-2 s-1)"])
CO2_ambient = list(df["CO2 - Ambient (umol mol-1)"])
CO2_elevated = list(df["CO2 - Elevated (umol mol-1)"])

# Ozone Exposure
fig, ax = plt.subplots()   # Create plot
ax.minorticks_on()         # Add minor tick marks
bin_n = np.linspace(0, 30, 31)   # Returns integer 1 over the interval 0 to 30

ax.hist(AOT40_elevated, bin_n, label="Elevated")   # Generate a histogram using AOT40_elevated as data
ax.hist(AOT40_ambient, bin_n, label="Ambient")     # Generate a histogram using AOT40_ambient as data

ax.set_title("Ozone Exposure")           # Name the histogram title as "Ozone Exposure"
ax.set_xlabel("Ozone Exposure (ppb)")    # Name the x-axis as "Ozone Exposure (ppb)"
ax.set_ylabel("Frequency")               # Name the y-axis as "Frequency"
ax.legend()                              # Create legend for the Ambient and Elevated histograms

# Net Photosynthesis Rate
# Repeat the process 
fig, ax2 = plt.subplots()
ax2.minorticks_on()

ax2.hist(NPR_elevated, bins=30, label="Elevated")
ax2.hist(NPR_ambient, bins=30, alpha=0.86, label="Ambient")   # Generate a ambient histogram with transparaency of 0.86

ax2.set_title("Net Photosynthesis Rate")
ax2.set_xlabel("Net Photosynthesis Rate (umol m-2 s-1)")
ax2.set_ylabel("Frequency")
ax2.legend()

# Stomatal Conductance
# Repeat the process
fig, ax3 = plt.subplots()
ax3.minorticks_on()

ax3.hist(SC_elevated, bins=30, label="Elevated")
ax3.hist(SC_ambient, bins=30, alpha=0.86, label="Ambient")

ax3.set_title("Stomatal Conductance")
ax3.set_xlabel("Stomatal Conductance (mol m-2 s-1)")
ax3.set_ylabel("Frequency")
ax3.legend()

# Intercellular CO2 Concentrations
# Repeat the process 
fig, ax4 = plt.subplots()
ax4.minorticks_on()

ax4.hist(CO2_ambient, bins=30, label="Elevated")
ax4.hist(CO2_elevated, bins=30, alpha=0.86, label="Ambient")

ax4.set_title("Intercellular CO2 Concentrations")
ax4.set_xlabel("Intercellular CO2 Concentrations (umol mol-1)")
ax4.set_ylabel("Frequency")
ax4.legend()
```
